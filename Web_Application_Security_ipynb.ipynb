**Web Application Security**

1. **Introduction**

Web applications have become an important part of modern digital services. Applications such as banking portals, e-commerce websites, social media platforms, educational portals, healthcare systems, and cloud applications process large amounts of sensitive information.

Because web applications are accessible through the Internet, they are exposed to various security threats. Attackers may attempt to access unauthorized information, manipulate application functionality, steal user sessions, inject malicious input, or exploit weaknesses in application design.

Web Application Security is the practice of protecting web applications, users, servers, databases, APIs, and sensitive information from unauthorized access and malicious activity.

The main objectives of web application security are:

* Confidentiality of information
* Integrity of application data
* Availability of services
* Authentication of users
* Authorization of actions
* Protection of sensitive information
* Prevention of common web attacks
* Secure application development

4. **HTTP Request and Response**

HTTP is one of the fundamental protocols used by web applications.

A client sends an HTTP request, and the server returns an HTTP response.

**Example Request**

* GET /profile HTTP/1.1
* Host: example.com
* Cookie: session=abc123

**Example Response**

* HTTP/1.1 200 OK
* Content-Type: text/html

Understanding HTTP traffic is important because many web security issues can be identified by analyzing requests and responses.

5. **HTTP Methods**

Common HTTP methods include:

| Method | Purpose                      |
| ------ | ---------------------------- |
| GET    | Retrieve information         |
| POST   | Submit/create information    |
| PUT    | Update information           |
| PATCH  | Partially update information |

Security testing should consider whether each HTTP method is properly authorized and handled by the application.



6. **HTTP Status Codes**

HTTP status codes indicate the result of a request.

**2xx – Success**

* 200 OK
* 201 Created
* 204 No Content


**3xx – Redirection**

* 301 Moved Permanently
* 302 Found
* 304 Not Modified

**4xx – Client Errors**

* 400 Bad Request
* 401 Unauthorized
* 403 Forbidden
* 404 Not Found

**5xx – Server Errors**

* 500 Internal Server Error
* 502 Bad Gateway
* 503 Service Unavailable

Security testers can use these responses to understand how an application behaves under different conditions.


7. **Cookies**

Cookies are small pieces of information stored by the browser and associated with a website.

They may be used for:

* Session management
* User preferences
* Authentication
* Shopping carts
* Tracking

Important cookie security attributes include:

* Secure
* HttpOnly
* SameSite

**Secure**

The cookie should only be transmitted over HTTPS.

**HttpOnly**

Helps prevent client-side scripts from directly accessing the cookie.

**SameSite**

Controls when cookies are sent in cross-site requests and can help reduce certain cross-site attack scenarios.


8. **Sessions**

11. **OWASP Top 10**

The OWASP Top 10 is a widely used awareness document that highlights important categories of web application security risks.

The current OWASP Top 10 categories are:

1. Broken Access Control
2. Cryptographic Failures
3. Injection
4. Insecure Design
5. Security Misconfiguration
6. Vulnerable and Outdated Components
7. Identification and Authentication Failures
8. Software and Data Integrity Failures
9. Security Logging and Monitoring Failures
10. Server-Side Request Forgery (SSRF)

12. **Broken Access Control**

Broken Access Control occurs when users can perform actions or access resources beyond their intended permissions.

**Example**

A normal user should only access:

/profile/my-account

but the application incorrectly allows access to another user's resources.

**Possible Impact**

* Unauthorized data access
* Account modification
* Exposure of sensitive information
* Privilege escalation

**Prevention**

* Implement server-side authorization
* Follow least privilege
* Deny access by default
* Validate object ownership
* Test authorization controls regularly

13. **Cryptographic Failures**

Cryptographic failures occur when sensitive information is not adequately protected.

**Examples include:**

* Weak encryption
* Plaintext transmission
* Poor password storage
* Exposed sensitive information
* Improper key management
**Prevention**

* Use HTTPS/TLS
* Use strong password hashing
* Protect encryption keys
* Avoid storing unnecessary sensitive data
* Use established cryptographic libraries

14. **Injection**

Injection vulnerabilities occur when untrusted input is interpreted as commands or queries.

**Examples include:**

* SQL injection
* Command injection
* LDAP injection
* Expression-language injection

**Example Concept**

Unsafe application logic might combine user input directly with a database query.

The secure approach is to use:

* Parameterized queries
* Prepared statements
* Input validation
* Appropriate escaping
* Least-privilege database accounts

Testing should only be performed against the intentionally vulnerable lab environment.

15. **Insecure Design**

Insecure Design refers to security weaknesses created during application architecture or business-logic design.

**Examples:**

* No rate limiting for sensitive operations
* Missing authorization checks
* Weak account recovery design
* No protection against automated abuse

**Prevention**

Security should be considered during the design stage rather than only after development.

Useful practices include:

* Threat modeling
* Abuse-case analysis
* Security requirements
* Secure architecture reviews
* Security testing

16. **Security Misconfiguration**

Security misconfiguration can occur when systems or applications use insecure settings.

**Examples:**

* Default credentials
* Unnecessary services
* Detailed error messages
* Debug mode enabled
* Incorrect permissions
* Missing security headers

**Prevention**

* Remove unnecessary components
* Disable debug functionality in production
* Use secure configuration baselines
* Regularly review configurations
* Keep software updated

17. **Vulnerable and Outdated Components**

Modern applications often depend on external libraries and frameworks.

Using outdated components can introduce known security vulnerabilities.

**Prevention**

* Maintain an inventory of dependencies
* Update dependencies regularly
* Monitor security advisories
* Remove unused packages
* Use automated dependency scanning

18. **Identification and Authentication Failures**

Authentication weaknesses can allow attackers to compromise accounts.

**Examples:**

* Weak passwords
* Poor session management
* Missing MFA where appropriate
* Weak account recovery
* Credential stuffing exposure
* Improper logout behavior

**Prevention**

* Strong authentication mechanisms
* MFA/passkeys where appropriate
* Secure password storage
Rate limiting
* Secure session management
* Account recovery protections

19. **Software and Data Integrity Failures**

Applications sometimes depend on software updates, packages, plugins, or data from external sources.

If integrity is not properly verified, malicious or unauthorized modifications may be introduced.

**Prevention**

* Verify software sources
* Secure CI/CD pipelines
* Protect build systems
* Verify package integrity
* Restrict deployment permissions

20. **Security Logging and Monitoring Failures**

Security logs help organizations detect suspicious activity.

**Important events may include:**

* Login failures
* Privilege changes
* Password changes
* Account modifications
* Administrative actions
* Security policy violations

Logs should be protected against unauthorized modification and should avoid storing unnecessary sensitive information.

21. **Server-Side Request Forgery**

SSRF occurs when a server is tricked into making requests to unintended destinations.

Potentially affected systems may include:

Web Application

      ↓

Internal Services

      ↓

Cloud/Internal Resources

**Prevention**

* Validate destination URLs
* Use allowlists where possible
* Restrict outbound network access
* Block access to sensitive internal resources
* Monitor unusual outbound requests

Testing should remain within an authorized lab.

23. **Vulnerability Severity**

Security findings can be categorized according to their potential impact.

| Severity      | Description                                 |
| ------------- | ------------------------------------------- |
| Critical      | Severe compromise with major consequences   |
| High          | Significant security impact                 |
| Medium        | Moderate impact or limited exploitation     |
| Low           | Minor security weakness                     |
| Informational | Observation or best-practice recommendation |



For a professional report, severity should be justified based on the actual evidence and impact rather than simply assigned based on the vulnerability name.
