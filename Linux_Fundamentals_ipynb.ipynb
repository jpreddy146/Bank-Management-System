**Linux Permissions, Users & Processes**

1. **Linux Permissions**

Run:

* *ls -l*

You may see something like:

* *-rw-r--r--  user user  notes.txt*

The permissions are generally divided into:

* Owner
* Group
* Others

And the basic permissions are:

* r = read
* w = write
* x = execute

Example

* -rwxr-xr--

Conceptually:

* Owner  → rwx
* Group  → r-x
* Others → r--

2. **chmod**


chmod changes permissions.

**Example:**

* chmod 600 private.txt

This means the owner gets read/write access while group and others don't get those permissions.

**Another example:**

* chmod 755 script.sh

**Reconnaissance**

**What is Reconnaissance?**

Reconnaissance means collecting information about a target before security testing.

Think about a security guard inspecting a building:

* What is the building?
* Who owns it?
* What entrances exist?
* What services are visible?

In cybersecurity, similar questions are asked about systems and networks.

**Passive vs Active Reconnaissance**


**Passive Reconnaissance**

You collect information without directly interacting with the target's systems in a probing/scanning manner.

**Examples:**

* Public websites
* Public documentation
* Search engines
* Public DNS information
* Public registration information * where available

**Example:**

Researching a company's publicly available technology information.

**Active Reconnaissance**

You directly interact with the target.

Examples include:

* Port scanning
* Service enumeration
* Network probing

For learning, perform these only against your controlled lab.

**Easy Difference**

| Passive                        | Active                        |
| ------------------------------ | ----------------------------- |
| Indirect information gathering | Direct interaction            |
| Lower chance of detection      | More likely to be logged      |
| Public information             | Network/service probing       |
| Example: public DNS records    | Example: authorized Nmap scan |


**Nmap & Wireshark**

1. **Nmap**


**Nmap = Network Mapper**

It is used for network discovery and security auditing.

You can use it to understand:

* Which hosts are reachable
* Which ports are open
* Which services are available
* Sometimes which service versions are detected

**2. Wireshark**

Wireshark is a network protocol analyzer.

It allows you to inspect network traffic captured from an interface.

You can learn about:

* IP addresses
* TCP/UDP
* DNS
* HTTP
* TLS
* Packet structure

**Burp Suite + OWASP ZAP + Recon Lab**

**1. Burp Suite**

Burp Suite is a web security testing platform.

It can help security testers understand HTTP/HTTPS requests and responses.

Basic concept:

Browser

   ↓

Burp Proxy

   ↓

Web Application


Instead of:

Browser → Website

you can observe the traffic through the proxy in your controlled lab.

**Learn these Burp concepts:**

* Proxy
* HTTP requests
* HTTP responses
* Headers
* Parameters
* Cookies
* Repeater

For your first lab, focus mainly on Proxy and understanding requests/responses.
